# <font color='#FFE15D'>**Predict $ \sin(x) $ / $ \cos(x) $**</font>

## **🔸Import**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## **🔸Functions**

```python
def predict_sin(x, weights):

    x = np.asarray(x)

    features = np.array([
        x**i for i in range(len(weights))
    ])

    return np.dot(features, weights)
```

```python
def predict_sin_cos(x, weights):

    if not 0 <= x <= 180:
        raise ValueError("x must be between 0 and 180")

    sin_value = predict_sin(x / 180, weights)

    if x <= 90:
        cos_value = predict_sin((90 - x) / 180, weights)
    else:
        cos_value = -predict_sin((x - 90) / 180, weights)

    return sin_value, cos_value
```

## **🔸Use Model**

```python
weights = np.load("polynomial-linear-regression.npy")

print(weights)
print(weights.shape)
```

## **🔸Input your degree**

```python
user_Input = float(input("Enter your input: "))

sin_pred, cos_pred = predict_sin_cos(user_Input, weights)

print(
    f"Degree: {user_Input:6.2f} | "
    f"Sin: {sin_pred[0]:.4f} | "
    f"Cos: {cos_pred[0]:.4f}"
)
```
